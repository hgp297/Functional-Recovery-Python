## Manual comparison plot

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# -----------------------------
# Data
# -----------------------------
data = [
    ("haseltonRCMF_4story", "Reoccupancy", 74.0, 73.9),
    ("haseltonRCMF_4story", "Functional", 108.1, 108.2),
    ("haseltonRCMF_4story", "Full", 127.3, 127.0),

    ("haseltonRCMF_12story", "Reoccupancy", 125.5, 125.1),
    ("haseltonRCMF_12story", "Functional", 154.1, 153.5),
    ("haseltonRCMF_12story", "Full", 177.2, 176.6),

    ("ICSB", "Reoccupancy", 536.2, 536.7),
    ("ICSB", "Functional", 574.1, 574.0),
    ("ICSB", "Full", 574.3, 574.3),

    ("RCSW_1story", "Reoccupancy", 1.7, 1.8),
    ("RCSW_1story", "Functional", 30.7, 30.8),
    ("RCSW_1story", "Full", 45.3, 45.4),
]

df = pd.DataFrame(data, columns=["Case", "Target", "Mean_Mat", "Mean_Py"])

# -----------------------------
# Plot setup
# -----------------------------
cases = df["Case"].unique()
targets = df["Target"].unique()

x = np.arange(len(cases))
width = 0.12  # bar width

fig, ax = plt.subplots(figsize=(10, 5))

# color mapping
py_color_2 = '#4B8BBE'
py_color_1 = '#FFD43B'

mat_color_1 = '#5778a4'
mat_color_2 = '#e49444'

# -----------------------------
# Plot grouped bars
# -----------------------------
for i, target in enumerate(targets):
    sub = df[df["Target"] == target]

    # align by case order
    mat_vals = [sub[sub["Case"] == c]["Mean_Mat"].values[0] for c in cases]
    py_vals  = [sub[sub["Case"] == c]["Mean_Py"].values[0] for c in cases]

    offset = (i - len(targets)/2) * width * 2

    ax.bar(x + offset - width/2, mat_vals, width,
           label=f"MATLAB implementation" if i == 0 else "",
           color=mat_color_1, alpha=0.7, edgecolor=mat_color_2)

    ax.bar(x + offset + width/2, py_vals, width,
           label=f"Python implementation" if i == 0 else "",
           color=py_color_1, alpha=0.7, edgecolor=py_color_2)
    
    # -----------------------------
    # Annotate recovery target ABOVE each cluster
    # -----------------------------
    for j, c in enumerate(cases):
        ymax = max(mat_vals[j], py_vals[j])

        ax.text(
            x[j] + offset,
            ymax + ymax * 0.03,   # small vertical offset
            target,
            ha="center",
            va="bottom",
            fontsize=10,
            rotation=0
        )


# -----------------------------
# Styling
# -----------------------------
ax.set_xticks(x)
ax.set_xticklabels(cases)
ax.set_ylabel("Mean Recovery Time (days)")
ax.set_title("Mat vs Py Mean Recovery Times by Case and Target")

ax.legend(ncol=2)
ax.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

# Deep dive plots for one example

### Histogram tool

```python
import os
import json
import numpy as np
import seaborn as sb
import matplotlib.pyplot as plt

py_color_2 = '#4B8BBE'
py_color_1 = '#FFD43B'

mat_color_1 = '#5778a4'
mat_color_2 = '#e49444'

plt.rcdefaults()

plt.rcParams['font.size'] = 12

def plt_histograms(recov_py, recov_mat):
    
    '''Plot all realizations of building level recovery as a histogram
    
    Parameters
    ----------
    recovery: dictionary
      dictionary containing building-level recovery data from the 
      functional recovery assessment
     
    plot_dir: str
      Save directory for plots. Plots will save directly to this location as
      png files.
    
    Returns
    -------'''
    
    # plt.rcParams["font.family"] = "Times New Roman"
    
    # ## Initial Setup
    # if os.path.exists(plot_dir) == False:
    #     os.mkdir(plot_dir)
    # plot_dir=plot_dir +'/'
    
    
    ## Plot Histograms
    # fnc_states = list(recov_py.keys())
    fnc_states = ['functional']
    for fs in range(len(fnc_states)):
        
        fig, ax = plt.subplots(figsize=(6,4))
        sb.histplot(data = recov_py[fnc_states[fs]]['building_level']['recovery_day'], alpha=1, color=py_color_1, edgecolor=py_color_2, label='ATCPy')
        sb.histplot(data = recov_mat[fnc_states[fs]]['building_level']['recovery_day'], alpha=0.6, color=mat_color_1, edgecolor=mat_color_2, label='ATCMat')

        max_ht = max([p.get_height() for p in ax.patches])

        py_mean = np.mean(recov_py[fnc_states[fs]]['building_level']['recovery_day'])
        mat_mean = np.mean(recov_mat[fnc_states[fs]]['building_level']['recovery_day'])
        ax.axvline(py_mean, color=py_color_1, linestyle='--')
        ax.axvline(mat_mean, color=mat_color_1, linestyle='--')

        x1 = min(py_mean, mat_mean)
        x2 = max(py_mean, mat_mean)

        ax.annotate(
            '', 
            xy=(x1, max_ht), xytext=(x2, max_ht),
            arrowprops=dict(arrowstyle='<->', shrinkA=0, shrinkB=0, 
            color='red'),
        )

        ax.annotate(
            r'$\Delta$ of mean', color='red',
            xy=(x1, max_ht), xytext=(x1*1.2, max_ht),
            arrowprops=dict(arrowstyle='-',
            color='red'),
        )
        

        fnc_lab = fnc_states[fs][0].upper() + fnc_states[fs][1:len(fnc_states[fs])]+' Recovery'        
        ax.set_xlabel(fnc_lab + ' Time (days)')
        ax.set_ylabel('Number of Realizatons')
        ax.legend()
        fig.tight_layout()
        # plt.savefig(plot_dir + fnc_states[fs] + '.png', dpi=300)
        
    return fig
```

### Systems breakdown tool

```python
def plt_demo_breakdown(recov_py, recov_mat):
    '''Plot the time and percent of realizations that each system and/or 
    component is impeding function as a heatmap and as lineplots
    
    Parameters
    ----------
    recovery: dictionary
     dictionary containing recovery breakdown data from the functional
     recovery assessment
     
    plot_dir: str
     Save directory for plots. Plots will save directly to this location as
     png files.
    
    Returns
    -------'''
    
    # ## Initial Setup
    # if os.path.exists(plot_dir) == False:
    #     os.mkdir(plot_dir)
    # plot_dir=plot_dir +'/'
    
    perform_targ_days = recov_py['reoccupancy']['breakdowns']['perform_targ_days']
    perform_targ_labs = ['Immediately', '>3 Days', '>7 Days', '>2 Weeks', '>1 Month', '>2 Months', '>3 Months', '>4 Months', '>6 Months', '>9 Months', '>1 Year', ]
    
    # Maximum day for recovery amongs all tenant units and all realizations
    recovery_day_max = perform_targ_days[-1]
   
    if recovery_day_max <= 365:
        perform_targ_labs = perform_targ_labs # Number of days for each performance target stripe
    if recovery_day_max > 365:
        num_years = (int(np.floor((recovery_day_max)/365)))
        
        # If number of complete years is more than 2
        for yr in range(num_years-1):
            quarters =[1,2,3]
            for qtr in quarters:
                if yr == 0:
                    perform_targ_labs.append('>' + str(yr+1) +' Year & ' + str(qtr*3)+ ' months')
                if yr > 0:
                    perform_targ_labs.append('>' + str(yr+1) +' Years & ' + str(qtr*3)+ ' months')    
            perform_targ_labs.append('>' + str(yr+2) +' Years')
        
        # For final incomplete year
        num_quarters = int(np.floor((recovery_day_max - num_years*365)/90))
        compl_yr = perform_targ_labs[-1]
        for qtr in range(num_quarters):
            perform_targ_labs.append(compl_yr + ' & '+str((qtr+1)*3) + ' Months')

        perform_targ_labs.append(perform_targ_labs[-1] + ' & ' + str(int(perform_targ_days[-1] - perform_targ_days[-2])) + ' Days')    

    
    var = ['component_breakdowns', 'system_breakdowns']
    labs = ['comp_names', 'system_names']
    plt_ht = [10, 8]
    
    ## Plot Heatmaps
    fnc_states = list(recov_py.keys())
    fig_list = []
    for fs in range(len(fnc_states)):
        
        perform_targ_days = recov_py[fnc_states[fs]]['breakdowns']['perform_targ_days']
        perform_targ_labs = ['Immediately', '>3 Days', '>7 Days', '>2 Weeks', '>1 Month', '>2 Months', '>3 Months', '>4 Months', '>6 Months', '>9 Months', '>1 Year', ]
        
        #FZ# Maximum day for recovery amongs all tenant units and all realizations
        recovery_day_max = perform_targ_days[-1]
       
        if recovery_day_max <= 365:
            perform_targ_labs = perform_targ_labs # Number of days for each performance target stripe
        if recovery_day_max > 365:
            num_years = (int(np.floor((recovery_day_max)/365)))
            
            # If number of complete years is more than 2
            for yr in range(num_years-1):
                quarters =[1,2,3]
                for qtr in quarters:
                    if yr == 0:
                        perform_targ_labs.append('>' + str(yr+1) +' Year & ' + str(qtr*3)+ ' months')
                    if yr > 0:
                        perform_targ_labs.append('>' + str(yr+1) +' Years & ' + str(qtr*3)+ ' months')    
                perform_targ_labs.append('>' + str(yr+2) +' Years')
            
            # For final incomplete year
            num_quarters = int(np.floor((recovery_day_max - num_years*365)/90))
            compl_yr = perform_targ_labs[-1]
            for qtr in range(num_quarters):
                perform_targ_labs.append(compl_yr + ' & '+str((qtr+1)*3) + ' Months')
    
            perform_targ_labs.append(perform_targ_labs[-1] + ' & ' + str(int(perform_targ_days[-1] - perform_targ_days[-2])) + ' Days')        
        
        # system breakdown for line plots
        sys_bkdn_py = np.array(recov_py[fnc_states[fs]]['breakdowns'][var[1]])
        sys_bkdn_mat = np.array(recov_mat[fnc_states[fs]]['breakdowns'][var[1]])
        systs = recov_py[fnc_states[fs]]['breakdowns'][labs[1]]
        # no. of systems in one plot (total 3 plots)
        num_systems = 1
        
        # Plot 1
        fig, ax = plt.subplots(figsize=(6,6))
        for syst in range(num_systems):
            plt.plot(perform_targ_days, sys_bkdn_py[syst], marker ='o', label='System recovery, Python', color=py_color_1)
            plt.plot(perform_targ_days, sys_bkdn_mat[syst], marker ='o', label='System recovery, MATLAB', color=mat_color_1)
            plt.fill_between(perform_targ_days, sys_bkdn_py[syst], sys_bkdn_mat[syst], color='red', alpha=0.3)
            y = list(zip(sys_bkdn_py[syst], sys_bkdn_mat[syst]))
            plt.plot((perform_targ_days,perform_targ_days),([i for (i,j) in y], [j for (i,j) in y]),c='red')
        differences = [mat - py for mat, py in y]
        max_diff = max(differences)
        max_idx = differences.index(max_diff)

        x1 = perform_targ_days[max_idx]
        y_val = y[max_idx][0]*.98

        ax.annotate(
            r'P95 of system $\Delta$', color='red',
            xy=(x1, y_val), xytext=(x1*1.2, y_val),
            arrowprops=dict(arrowstyle='->',
            color='red'),
        )

        # plt.ylim([0,1])
        plt.xlabel('Recovery Time After Earthquake (Days)')
        plt.ylabel('Fraction of Realizations Affecting Building')
        # plt.title('System Recovery Breakdown')
        plt.legend()
        plt.grid()
        fig.tight_layout()

        fig_list.append(fig)

    return fig_list
```

### Trajectory tool

```python
def plt_recovery_trajectory(recovery, full_repair_time):
    '''Plot mean recovery trajectories
    
    Parameters
    ----------
    recovery: dictionary
      dictionary containing recovery trajectory data from the functional
      recovery assessment
     
    full_repair_time: array [num reals x 1] 
      simulated realization of full repair time 
    
    plot_dir: str
      Save directory for plots. Plots will save directly to this location as
      png files.
    
    Returns
    -------'''
     
    
    # ## Initial Setup
    # if os.path.exists(plot_dir) == False:
    #     os.mkdir(plot_dir)
    # plot_dir=plot_dir +'/'
    realization_color = 'lightgray'
    mean_color = '#002145'
    stdv_color = '#0077C8'
    
    

    # Calculate mean recovery times
    reoc = np.mean(np.array(recovery['reoccupancy']['recovery_trajectory']['recovery_day']), axis=0)
    func = np.mean(np.array(recovery['functional']['recovery_trajectory']['recovery_day']), axis=0)
    med = np.median(np.array(recovery['functional']['recovery_trajectory']['recovery_day']), axis=0)
    per_10 = np.percentile(np.array(recovery['functional']['recovery_trajectory']['recovery_day']), 84, axis=0)
    per_90 = np.percentile(np.array(recovery['functional']['recovery_trajectory']['recovery_day']), 16, axis=0)
    full = np.mean(full_repair_time)
    level_of_repair = recovery['functional']['recovery_trajectory']['percent_recovered']
    
    # Plot Recovery Trajectory
    mean_fig = plt.figure(figsize=(6,4)) 
    plt.plot(reoc, level_of_repair,'r-', linewidth = 1.5, label = 'Re-Occupancy') 
    plt.plot(func, level_of_repair,'b-', linewidth = 1.5, label= 'Functional') 
    plt.plot([full, full], [0, 1],'k-', linewidth = 1.5, label= 'Fully Repaired') 
    plt.xlim([0,np.ceil((full+1)/10)*10])
    plt.xlabel('Days After Earthquake')
    plt.ylabel('Fraction of Floor Area')
    plt.legend()
    # plt.title('Mean Recovery Trajectories')
    plt.grid()
    mean_fig.tight_layout()
    
    
    func_fig = plt.figure(figsize=(6,4))
    for i in range(len(recovery['functional']['recovery_trajectory']['recovery_day'])):
        plt.plot(recovery['functional']['recovery_trajectory']['recovery_day'][i] , level_of_repair, color = 'lightgrey', linewidth = 0.5, alpha=0.2)
    plt.plot(med, level_of_repair, color=mean_color, linewidth = 2.5, label= 'Median')
    plt.plot(per_10, level_of_repair,'--', color=stdv_color, linewidth = 1.5, label= '16-84th percentiles')
    plt.plot(per_90, level_of_repair,'--', color=stdv_color, linewidth = 1.5)
    plt.xlim([0,np.ceil((max(max(recovery['functional']['recovery_trajectory']['recovery_day'])))/10)*10])
    # plt.title('Functional Recovery Trajectories')
    plt.xlabel('Days After Earthquake')
    plt.ylabel('Fraction of Floor Area')
    plt.legend()
    plt.grid()    
    
    func_fig.tight_layout()

    med = np.median(np.array(recovery['reoccupancy']['recovery_trajectory']['recovery_day']), axis=0)
    per_10 = np.percentile(np.array(recovery['reoccupancy']['recovery_trajectory']['recovery_day']), 10, axis=0)
    per_90 = np.percentile(np.array(recovery['reoccupancy']['recovery_trajectory']['recovery_day']), 90, axis=0)
    full = np.mean(full_repair_time)
    level_of_repair = recovery['reoccupancy']['recovery_trajectory']['percent_recovered']

    recov_fig = plt.figure(figsize=(6,4))
    for i in range(len(recovery['reoccupancy']['recovery_trajectory']['recovery_day'])):
        plt.plot(recovery['reoccupancy']['recovery_trajectory']['recovery_day'][i] , level_of_repair, color = 'lightgrey', linewidth = 0.5, alpha=0.2)
    plt.plot(med, level_of_repair, color=mean_color, linewidth = 2.5, label= 'Median')
    plt.plot(per_10, level_of_repair,'--', color=stdv_color, linewidth = 1.5, label= '16-84th percentiles')
    plt.plot(per_90, level_of_repair,'--', color=stdv_color, linewidth = 1.5)
    plt.xlim([0,np.ceil((max(max(recovery['reoccupancy']['recovery_trajectory']['recovery_day'])))/10)*10])
    # plt.title('Reoccupancy Trajectories')
    plt.xlabel('Days After Earthquake')
    plt.ylabel('Fraction of Floor Area')
    plt.legend()
    plt.grid()    
    recov_fig.tight_layout()

    
    return mean_fig, recov_fig, func_fig
```

### Load tool

```python

def load_results(res_dir):
    f = open(os.path.join(res_dir))
    functionality= json.load(f)

    if 'functionality' in functionality:
        functionality = functionality['functionality']

    recovery = functionality['recovery']
    impede = functionality['impeding_factors']['breakdowns']['full']
    schedule = functionality['building_repair_schedule']
    workers = functionality['worker_data']
    full_repair_time = np.nanmax(np.array(schedule['full']['repair_complete_day']['per_story']))

    result_dict = {
        'recovery': recovery,
        'impede': impede,
        'schedule': schedule,
        'workers': workers,
        'full_repair_time': full_repair_time
    }
    return result_dict
```
